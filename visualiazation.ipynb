```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv("C:/Terraform/TRUPTI/The  developer arena/week4-Sales -analysis/sales_data (2).csv")

# Count sales for each product
product_sales = df.groupby("Product")["Total_Sales"].sum()

# Create bar chart
product_sales.plot(kind="bar")

plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.show()
```
